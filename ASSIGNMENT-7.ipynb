### a) Different ways to iterate over rows in Pandas DataFrame

```python
import pandas as pd

df = pd.DataFrame({
    "name": ["Aarav", "Priya", "Rahul"],
    "age": [25, 30, 35],
    "city": ["Mumbai", "Delhi", "Bengaluru"]
})

for index, row in df.iterrows():
    print(index, row["name"], row["age"], row["city"])

for row in df.itertuples(index=True):
    print(row.Index, row.name, row.age, row.city)

for row in df.values:
    print(row[0], row[1], row[2])

for row in df.to_dict(orient="records"):
    print(row["name"], row["age"], row["city"])

df["age_plus_5"] = df["age"].apply(lambda x: x + 5)

df["double_age"] = df["age"] * 2

print(df)
```

### b) Selecting rows in pandas DataFrame based on conditions

```python
import pandas as pd

data = {
    "name": ["Naman", "Devesh", "Garvit"],
    "age": [21, 18, 25]
}

df = pd.DataFrame(data)

result = df[df["age"] > 20]

print(result)
```

### c) Select any row from a DataFrame using iloc[]

```python
import pandas as pd

data = {
    "name": ["Hitesh", "Garv", "Devesh"],
    "age": [21, 18, 25]
}

df = pd.DataFrame(data)



print(df.iloc[1])
```

### d) Limited rows selection with given column

```python
import pandas as pd

data = {
    "name": ["Shekhar", "Shan", "Garv"],
    "age": [21, 22, 23],
    "City": ["Alwar", "Bhopal", "Jaipur"]
}

df = pd.DataFrame(data)

print(df.loc[0:1, ["name", "City"]])
```

### e) Drop rows from the DataFrame based on condition

```python
import pandas as pd

data = {
    "name": ["Devesh", "Lakshay", "Dev"],
    "age": [22,23,20],
    "City": ["Alwar", "Pune", "Delhi"]
}

df = pd.DataFrame(data)

df=df[df["name"]=="Dev"]
print(df)
```

### f) Insert row at given position in Pandas DataFrame

```python
import pandas as pd

df = pd.DataFrame({
    "name": ["Aarav", "Priya", "Rahul"],
    "age": [25, 30, 35],
    "city": ["Mumbai", "Delhi", "Bengaluru"]
})

new_row = pd.DataFrame({
    "name": ["Sneha"],
    "age": [28],
    "city": ["Jaipur"]
})

df = pd.concat([
    df.iloc[:1],
    new_row,
    df.iloc[1:]
]).reset_index(drop=True)

print(df)
```
