```python
import pandas as pd

df1 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Name': ['A', 'B', 'C']
})

df2 = pd.DataFrame({
    'ID': [1, 2, 4],
    'Marks': [90, 85, 70]
})

print("DataFrame 1:")
print(df1)

print("\nDataFrame 2:")
print(df2)

# Inner Merge
inner_merge = pd.merge(df1, df2, on='ID', how='inner')

print("\nInner Merge:")
print(inner_merge)

# Left Join
left_join = pd.merge(df1, df2, on='ID', how='left')

print("\nLeft Join:")
print(left_join)

# Right Join
right_join = pd.merge(df1, df2, on='ID', how='right')

print("\nRight Join:")
print(right_join)

# Index-Based Join
df1_index = df1.set_index('ID')
df2_index = df2.set_index('ID')

index_join = df1_index.join(df2_index)

print("\nIndex-Based Join:")
print(index_join)

# Multiple Keys Merge
df3 = pd.DataFrame({
    'ID': [1, 1, 2],
    'Dept': ['CS', 'IT', 'CS'],
    'Name': ['A', 'B', 'C']
})

df4 = pd.DataFrame({
    'ID': [1, 2, 1],
    'Dept': ['CS', 'CS', 'IT'],
    'Marks': [90, 85, 88]
})

multi_merge = pd.merge(df3, df4, on=['ID', 'Dept'])

print("\nMerge with Multiple Keys:")
print(multi_merge)
```

```python
import pandas as pd

employee_df1 = pd.DataFrame({
    'Emp_ID': [101, 102],
    'Employee': ['Rahul', 'Sneha']
})

employee_df2 = pd.DataFrame({
    'Emp_ID': [103, 104],
    'Employee': ['Aman', 'Priya']
})

salary_df = pd.DataFrame({
    'Emp_ID': [101, 102, 103, 104],
    'Salary': [45000, 50000, 47000, 52000]
})

combined_employees = pd.concat([employee_df1, employee_df2], ignore_index=True)

print("Combined Employee Data:")
print(combined_employees)

final_data = pd.merge(combined_employees, salary_df, on='Emp_ID')

print("\nMerged Employee and Salary Data:")
print(final_data)

employee_index = employee_df1.set_index('Emp_ID')
salary_index = salary_df.set_index('Emp_ID')

join_output = employee_index.join(salary_index)

print("\nResult using join():")
print(join_output)

merge_output = pd.merge(employee_df1, salary_df, on='Emp_ID')

print("\nResult using merge():")
print(merge_output)

# Difference between join() and merge()

# join()
# 1. Uses index by default
# 2. Simpler syntax
# 3. Called using df.join()
# 4. Mostly used for index-based combining

# merge()
# 1. Uses common columns for combining
# 2. More flexible and powerful
# 3. Called using pd.merge()
# 4. Mostly used for database-style joins
```
